# Creating variables and constants in tensor

```python
constant_tf=tf.constant([[1,2],[3,4]])
```

```python
constant_tf
```

```python
variable_tf=tf.Variable([[5,6],[1,3]])
```

```python
variable_tf
```

#Tensor Transpose

```python
transpose_tf=tf.transpose(constant_tf)
```

```python
transpose_tf
```

#Multiplication in tf

```python
multipl_mat=tf.matmul(constant_tf,transpose_tf)
```

```python
multipl_mat
```

```python
multipl_mat2=tf.multiply(constant_tf,transpose_tf)
```

```python
multipl_mat2
```

```python
tensor_transpose_float=tf.cast(constant_tf,tf.float32)
```

```python
determinant=tf.linalg.det(tensor_transpose_float)
```

```python
determinant
```

```python
inver = tf.linalg.inv(tensor_transpose_float)
```

```python
inver
```
